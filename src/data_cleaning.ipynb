```python
import pandas as pd

from Chord import Chord
from Rule import Rule
from collections import defaultdict
import pitchtypes as pt
from probabilistic_model import ProbabilisticModel
from itertools import product
import torch 
from Environment import Environment
from Helpers import get_possible_intervals, get_qualities, get_possible_rhs_from_str, get_applicable_rules, evaluate_tree_from_dict
from TreeNode import TreeNode, build_tree_from_dict
from DQN import train_on_batch, ReplayBuffer, train_model, DQN, DQNWithTransformer
import numpy as np
from sklearn.model_selection import train_test_split

import random
```

```python
data = pd.read_json('../eda/JazzHarmonyTreebank/treebank.json')
trees_data = data[data['trees'].notna()]
```

```python
train_data, test_data = train_test_split(trees_data, test_size=0.2, random_state=42)
```

```python
all_leaf_labels_train = [
    [i.chord.label for i in build_tree_from_dict(entry['trees'][0]['complete_constituent_tree']).get_leaves()]
    for _, entry in train_data.iterrows()
]

train_data['chords'] = all_leaf_labels_train

all_leaf_labels_test = [
    [i.chord.label for i in build_tree_from_dict(entry['trees'][0]['complete_constituent_tree']).get_leaves()]
    for _, entry in test_data.iterrows()
]

test_data['chords'] = all_leaf_labels_test
```

```python
train_data.to_csv('train_data.csv', index=False)
test_data.to_csv('test_data.csv', index=False)
```

```python
ccts = [i[0]['complete_constituent_tree'] for i in test_data['trees'] if len(i) > 0]
```
